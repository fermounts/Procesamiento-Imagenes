## Paso 3 — Cargar y mostrar la imagen

```
import cv2
import matplotlib.pyplot as plt

ruta = '/content/drive/MyDrive/Procesamiento-Imagenes/TP5/calle.jpg'

img = cv2.imread(ruta)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10, 8))
plt.imshow(img_rgb)
plt.title('Imagen original')
plt.axis('off')
plt.show()
```

## Paso 4 — Detección con YOLOv8

```
from ultralytics import YOLO
import matplotlib.pyplot as plt
import cv2

# Cargar modelo preentrenado YOLOv8
model = YOLO('yolov8n.pt')

# Correr detección
results = model(ruta)

# Mostrar imagen con detecciones
img_resultado = results[0].plot()
img_resultado_rgb = cv2.cvtColor(img_resultado, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10, 8))
plt.imshow(img_resultado_rgb)
plt.title('Detección YOLO')
plt.axis('off')
plt.show()
```

## Paso 5 — Ver objetos detectados

```
print('Objetos detectados:')
print('='*40)
for box in results[0].boxes:
    clase = results[0].names[int(box.cls)]
    confianza = float(box.conf)
    print(f'{clase}: {confianza:.2f}')
```

## Paso 6 — Comparativa original vs detección

```
fig, axes = plt.subplots(1, 2, figsize=(18, 8))
axes[0].imshow(img_rgb)
axes[0].set_title('Imagen original')
axes[0].axis('off')
axes[1].imshow(img_resultado_rgb)
axes[1].set_title('Detección YOLO')
axes[1].axis('off')
plt.tight_layout()
plt.show()
```
