## Paso 3 — Cargar y mostrar la imagen

```
import cv2
import matplotlib.pyplot as plt

# Ruta de la imagen en Google Drive
ruta = '/content/drive/MyDrive/Procesamiento-Imagenes/TP4/factura.jpeg'

img = cv2.imread(ruta)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10, 14))
plt.imshow(img_rgb)
plt.title('Imagen original')
plt.axis('off')
plt.show()
```

## Paso 4 — Preprocesamiento

```
import cv2
import matplotlib.pyplot as plt

# Escala de grises
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Reducir ruido
gray = cv2.medianBlur(gray, 3)

# Binarización
_, thresh = cv2.threshold(gray, 150, 255, cv2.THRESH_BINARY)

# Mostrar comparativa
fig, axes = plt.subplots(1, 3, figsize=(15, 6))
axes[0].imshow(img_rgb)
axes[0].set_title('Original')
axes[0].axis('off')
axes[1].imshow(gray, cmap='gray')
axes[1].set_title('Escala de grises')
axes[1].axis('off')
axes[2].imshow(thresh, cmap='gray')
axes[2].set_title('Binarizada')
axes[2].axis('off')
plt.tight_layout()
plt.show()
```

## Paso 5 — OCR con Tesseract

```
import pytesseract

config = r'--oem 3 --psm 6 -l spa'
texto = pytesseract.image_to_string(thresh, config=config)

print('Texto detectado:')
print('='*50)
print(texto)
```

## Paso 6 — Mejora con contraste (equalizeHist)

```
import pytesseract
import cv2

# Aplicar mejora de contraste
gray2 = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
gray2 = cv2.equalizeHist(gray2)
gray2 = cv2.medianBlur(gray2, 3)
_, thresh2 = cv2.threshold(gray2, 150, 255, cv2.THRESH_BINARY)

texto_mejorado = pytesseract.image_to_string(thresh2, config=config)

print('Texto detectado con mejora de contraste:')
print('='*50)
print(texto_mejorado)
```
